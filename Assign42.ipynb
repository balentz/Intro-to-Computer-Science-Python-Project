```python
def calculate_sum_from_input():
  print("Enter numbers separated by spaces or commas (e.g., 5, 10, 15):")
  user_input = input("Enter numbers: ")
  cleaned_input = user_input.replace(",", " ")
  entered = cleaned_input.split()
  numbers = []
  for t in entered:
    try:
      num = float(t)
      if num.is_integer():
        num = int(num)
      numbers.append(num)
    except ValueError:
      print("Skipping invalid entry:", t)
  if numbers:
    print("Numbers entered:", ", ".join(map(str, numbers)))
    total = sum(numbers)
    return total
  else:
    print("No valid numbers were entered.")
    return 0
total = calculate_sum_from_input()
print("Total Sum:", total)
```
