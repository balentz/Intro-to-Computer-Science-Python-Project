```python
grocery_list=[]

while True:
    print("\nGrocery List Manager")
    print("1. Add Item")
    print("2. Remove Item")
    print("3. Show Items")
    print("4. Exit")
    choice=(input("Enter your choice (1-4):"))
    if choice=="1":
        item=input("Enter the item to add:")
        grocery_list.append(item)
        print("Enter your choice:", choice)
        print("Enter the item to add:", item)
        print(item, "has been added to the grocery list")
    elif choice == "2":
        item=input("Enter the item to remove:")
        if item in grocery_list:
            grocery_list.remove(item)
            print("Enter your choice:", choice)
            print("Enter the item to remove:", item)
            print(item, "has been removed from the grocery list")
        else:
            print("Item not in the grocery list")
    elif choice == "3":
        print("Enter your choice:", choice)
        print("\nCurrent Grocery List:")
        if len(grocery_list)==0:
            print("Nothing has been added to the grocery list")
        for i in range(len(grocery_list)):

            print(i+1,grocery_list[i],sep=".")
    elif choice =="4":
        print("Exiting the Grocery List Manager. Goodbye!")
        break
    else:
        print("Enter your choice:", choice)
        print("Invalid Choice. Enter a number from 1 to 4")
```
